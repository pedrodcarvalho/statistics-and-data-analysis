```python
# Arrays
list1 = [1, 2, 3, 4, 5]
print(list1)

# Accessing elements
print(list1[0])
print(list1[0:3])  # Slicing

# Adding elements
list1.append(20)
print(list1)

list1.insert(2, 10)
print(list1)

# Concatenating lists
list2 = [2, 5, 6]
list1.extend(list2)
print(list1)

# Removing elements
for i in range(3):
    list1.pop()
print(list1)

# Creating a list of numbers
list(range(2, 31, 4))  # Create a list of numbers from 2 to 30 with a step of 4
```
