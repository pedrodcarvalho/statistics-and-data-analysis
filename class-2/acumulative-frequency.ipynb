```python
import math
import pandas as pd
```

```python
df = pd.read_excel('../datasets/project-1.xlsx')
df
```

```python
df.Idade.value_counts().sort_index()
```

```python
total = df.Idade.value_counts().sum()
```

```python
quantitative_data = pd.DataFrame(df)
quantitative_data = quantitative_data.groupby(
    'Idade').size().reset_index(name='counts')
```

```python
quantitative_data["Frequency %"] = round(
    quantitative_data['counts'] * 100 / total, 2)
quantitative_data
```

```python
quantitative_data["Acumulative Frequency %"] = quantitative_data['Frequency %'].cumsum()
quantitative_data
```
