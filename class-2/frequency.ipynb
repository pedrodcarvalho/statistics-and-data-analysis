```python
qualitative_data = pd.DataFrame(df)
qualitative_data = qualitative_data.groupby(
    'Escolaridade').size().reset_index(name='count')
qualitative_data
```

```python
total = qualitative_data['count'].sum()
total
```

```python
qualitative_data["Frequency %"] = round(
    qualitative_data['count'] * 100 / total, 2)
qualitative_data
```
