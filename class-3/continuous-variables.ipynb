1. Importar as bibliotecas "Pandas" e "Numpy" (Usaremos as abreviações pd e np, respectivamente, para faciliar a notação.
   Importe também a biblioteca Math.


```python
import pandas as pd
import numpy as np
import math
```

2. Defina a variável "arquivo" pelo nome do arquivo da planilha.
   Defina também a variável "coluna". Essa variável definirá qual a variável será considerada para a construção da tabela.


```python
arquivo = '../datasets/sodium-table.xlsx'
coluna = 'Sodium (mg)'
```

3. Leia a planilha no Phyton (para isso use o comando pd.read_excel(...) ). Dê um nome (nesse caso chamaremos de "df").


```python
df = pd.read_excel(arquivo)
```

4. Montando Tabela de classes com número de classes fixo. E ordenando as classes


Podemos escolher o número de classes ou utilizar o valor que vimos nos passos para montar tabelas com intervalos de classes (arquivo do exemplo do histograma) que é
**número de classes = raiz quadrada (número de dados)**


```python
numdados = len(df[coluna])
numdados
```

```python
numeroclasses = math.sqrt(numdados)
valorbins = round(numeroclasses)
valorbins
```

Montando a coluna de frequência absoluta


```python
frequencia = pd.cut(x=df[coluna], bins=valorbins,
                    include_lowest=True, right=False).value_counts()
frequencia.sort_index(ascending=True)
```

5. Montando Frequência Relativa


```python
Freqpercentual = pd.cut(x=df[coluna], bins=valorbins,
                        include_lowest=True, right=False).value_counts(normalize=True)
Freqpercentual.sort_index(ascending=True)
```

6. Montando as colunas da frequência acumulada e da frequência relativa acumulada


```python
freqacum = np.cumsum(frequencia.sort_index())
freqacum
```

```python
freqpercacum = np.cumsum(Freqpercentual.sort_index())
freqpercacum
```

7. Arrumando a Tabela


```python
dist_freq = round(pd.DataFrame({'Frequência': frequencia, 'Frequência Relativa (%)': Freqpercentual *
                  100, 'Freq.Acum.': freqacum, 'Freq.Rel.Acum. (%)': freqpercacum * 100}), 2)
print(dist_freq)
print(f'\n      Variável: {coluna}')
print('     *Limites superiores do intervalo de classes não incluídos')
```

8. Outra forma de montar indicando os limites das classes.Neste caso precisamos antes saber o valor mínimo e o valor máximo dos dados para que todos estejam em pelo menos uma das classes.


```python
valormin = df[coluna].min()
valormax = df[coluna].max()
print(valormin, valormax)
```

```python
classes = [0, 30000, 60000, 90000, 120000, 150000, 180000, 210000, 240000]
frequencia1 = (pd.cut(x=df[coluna], bins=classes,
               include_lowest=True, right=False)).value_counts()
frequencia1.sort_index(ascending=True)
```

```python
Freqpercentual1 = (pd.cut(x=df[coluna], bins=classes,
                   include_lowest=True, right=False)).value_counts(normalize=True)
Freqpercentual1.sort_index(ascending=True)
```

```python
dist_freq1 = round(pd.DataFrame(
    {'Frequência': frequencia1, 'Frequência Relativa (%)': Freqpercentual1*100}), 2)
tabela1 = dist_freq1.sort_index(ascending=True)
print(tabela1)
print(f'\n      Variável:{coluna}')
print('     *Limites superiores do intervalo de classes não incluídos')
```

**As colunas das frequências acumuladas são montadas da mesma forma descrita no item 6. e tabela ajustada no item 7.**
